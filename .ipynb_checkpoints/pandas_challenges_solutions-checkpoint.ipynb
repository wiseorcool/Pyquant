````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
df.head()
````

## 1. Data Exploration

**Q1. Show the first 7 rows.**

````
df.head(7)
````

**Q2. How many rows and columns?**

````
df.shape  # (rows, cols)
````

**Q3. Missing values per column.**

````
df.isna().sum().sort_values(ascending=False)
````

## 2. Selecting & Creating Columns

**Q4. Select only `Name` and `Fare`.**

````
df[['Name', 'Fare']].head()
````

**Q5. Create `FamilySize = SibSp + Parch + 1`.**

````
df['FamilySize'] = df['SibSp'].fillna(0) + df['Parch'].fillna(0) + 1
df[['SibSp','Parch','FamilySize']].head()
````

**Q6. Rename `Pclass` to `Class`.**

````
df_renamed = df.rename(columns={'Pclass': 'Class'})
df_renamed.columns
````

## 3. Filtering Rows

**Q7. Passengers older than 60.**

````
df[df['Age'] > 60][['PassengerId','Name','Age']].sort_values('Age')
````

**Q8. Fare greater than 100.**

````
df[df['Fare'] > 100][['PassengerId','Name','Fare','Pclass']].sort_values('Fare', ascending=False).head(10)
````

**Q9. Female passengers in 1st class.**

````
df[(df['Sex'] == 'female') & (df['Pclass'] == 1)][['PassengerId','Name','Sex','Pclass']].head(10)
````

## 4. GroupBy & Aggregation

**Q10. Average age by passenger class.**

````
df.groupby('Pclass')['Age'].mean().round(2)
````

**Q11. Survived vs not survived by gender.**

````
df.groupby(['Sex', 'Survived'])['PassengerId'].count().unstack(fill_value=0)
````

**Q12. Highest fare paid in each class.**

````
df.groupby('Pclass')['Fare'].max().sort_index()
````

## 5. Sorting & Ranking

**Q13. Top 5 youngest passengers.**

````
df[['PassengerId','Name','Age']].dropna(subset=['Age']).sort_values('Age').head(5)
````

**Q14. Top 5 highest fares.**

````
df[['PassengerId','Name','Fare']].sort_values('Fare', ascending=False).head(5)
````

## 6. Apply & Functions

**Q15. `AgeGroup`: 'Child' if Age < 18 else 'Adult'.**

````
df['AgeGroup'] = np.where(df['Age'] < 18, 'Child', 'Adult')
df[['Age','AgeGroup']].head(10)
````

**Q16. Name starts with 'A' (boolean).**

````
df['NameStartsWithA'] = df['Name'].str.strip().str.startswith('A')
df[['Name','NameStartsWithA']].head(10)
````

## 7. Advanced (Pivot, Merge, Resample)

**Q17. Pivot: average fare by Sex and Class.**

````
pivot = pd.pivot_table(df, values='Fare', index='Sex', columns='Pclass', aggfunc='mean')
pivot.round(2)
````

**Q18. Merge example with notes.**

````
notes = pd.DataFrame({
    'PassengerId': df['PassengerId'].sample(5, random_state=42).tolist(),
    'Note': ['VIP','Follow-up','Missing Age','Check Fare','Reconcile name']
})
merged = df.merge(notes, on='PassengerId', how='left')
merged[['PassengerId','Name','Note']].head(10)
````

**Q19. Resample monthly count if `Date` exists; otherwise demo.**

````
# If a Date column exists:
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    monthly = df.set_index('Date').resample('M').size()
else:
    # Demo: create a synthetic date range matching dataframe length
    dates = pd.date_range('1912-01-01', periods=len(df), freq='D')
    monthly = pd.Series(1, index=dates).resample('M').sum()
monthly.head()
````

## 8. Visualization

**Q20. Histogram of Age.**

````
plt.figure()
df['Age'].plot(kind='hist', bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
````

**Q21. Bar chart of survival rate by passenger class.**

````
survival_rate = df.groupby('Pclass')['Survived'].mean()
plt.figure()
survival_rate.plot(kind='bar')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.ylim(0,1)
plt.show()
````
