## Day 3 — Missing Data (`isna`, `fillna`, drop vs fill)
**Goal:** profile NA and apply group-wise fills where sensible.

```python
na_counts = df.isna().sum().sort_values(ascending=False)
print("Top missing columns:")
print(na_counts.head())

# Example: group-wise fill for Titanic 'age' by 'class' or 'sex'
if {'age','class'}.issubset(df.columns):
    df['age'] = df['age'].fillna(df.groupby('class')['age'].transform('median'))
elif {'sepal_length','species'}.issubset(df.columns):
    # Iris rarely has NA; demo imputation by species group (no-op if no NA)
    df['sepal_length'] = pd.to_numeric(df['sepal_length'], errors='coerce')
    df['sepal_length'] = df['sepal_length'].fillna(df.groupby('species')['sepal_length'].transform('median'))

# SELF-CHECK: chosen feature should have no NA now (if it existed)
if 'age' in df.columns:
    assert df['age'].isna().sum() == 0
else:
    assert df['sepal_length'].isna().sum() == 0

print("\n✅ Day 3 checks passed")
```

## Day 5 — Sorting & Ranking
**Goal:** multi-column sort and group-wise ranks.

```python
if {'class','fare'}.issubset(df.columns):
    df = df.sort_values(['class','fare'], ascending=[True, False]).copy()
    df['fare_rank_in_class'] = df.groupby('class')['fare'].rank('dense', ascending=False)
    # SELF-CHECK
    assert df['fare_rank_in_class'].ge(1).all()
elif {'species','sepal_length'}.issubset(df.columns):
    df = df.sort_values(['species','sepal_length'], ascending=[True, False]).copy()
    df['len_rank_in_species'] = df.groupby('species')['sepal_length'].rank('dense', ascending=False)
    # SELF-CHECK
    assert df['len_rank_in_species'].ge(1).all()

print("\n✅ Day 5 checks passed")
```

## Week 2 — Grouping & Reshaping
**Goal:** `groupby`, `agg`, `pivot_table`, `melt`.

```python
# GroupBy examples
if {'survived','sex'}.issubset(df.columns):
    summary = (df.groupby(['sex','class'], observed=True)
                 .agg(survival_rate=('survived','mean'),
                      avg_age=('age','mean'),
                      n=('survived','size'))
                 .reset_index())
    display(summary.head())
else:
    summary = (df.groupby('species', observed=True)
                 .agg(mean_len=('sepal_length','mean'),
                      mean_wid=('sepal_width','mean'),
                      n=('species','size'))
                 .reset_index())
    display(summary.head())

# Reshaping with pivot_table
if {'survived','class','sex'}.issubset(df.columns):
    pt = pd.pivot_table(df, index='class', columns='sex', values='survived', aggfunc='mean')
    display(pt)
else:
    pt = pd.pivot_table(df, index='species', values='sepal_length', aggfunc='mean')
    display(pt)

# Melt back to long
long = pt.reset_index().melt(id_vars=pt.index.name, var_name='key', value_name='value')
display(long.head())

print("\n✅ Week 2 basic grouping/reshaping done")
```
